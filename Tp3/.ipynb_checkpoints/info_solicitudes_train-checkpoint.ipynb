```python
# Usuarios distintos
print("Usuarios distintos:", solicitudes_train["ID_Usuario"].nunique())

# Filas duplicadas y solicitudes con ID repetido
print("Filas duplicadas:", solicitudes_train.duplicated().sum())

# Valores faltantes por columna
print("\nValores faltantes:")
print(solicitudes_train.isnull().sum())

# Período de las solicitudes
fechas = pd.to_datetime(solicitudes_train["Fecha_Solicitud"], errors="coerce")
print("\nPeríodo:", fechas.min().date(), "a", fechas.max().date())

# Máximo de solicitudes realizadas por un mismo usuario
solicitudes_por_usuario = solicitudes_train.groupby("ID_Usuario").size()
print("\nMáximo de solicitudes de un usuario:", solicitudes_por_usuario.max())
```
